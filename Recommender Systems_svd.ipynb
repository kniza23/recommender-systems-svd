```python
import numpy as np
import pandas as pd

def CF_userbased(u_id, K, data, is_rounded=True, sim_measure="pearson"):
    if u_id not in data.index:
        raise ValueError(f"L'utilisateur {u_id} n'existe pas dans la matrice d'utilité.")

    target_ratings = data.loc[u_id]

    non_zero_target = target_ratings[target_ratings > 0]
    mean_target = non_zero_target.mean() if len(non_zero_target) > 0 else 0

    similarities = {}

    for other_id in data.index:
        if other_id == u_id:
            continue
        other_ratings = data.loc[other_id]

        common_mask = (target_ratings > 0) & (other_ratings > 0)
        if not common_mask.any():
            continue

        u_common = target_ratings[common_mask]
        v_common = other_ratings[common_mask]

        if sim_measure == "pearson":
            mean_u = u_common.mean()
            mean_v = v_common.mean()

            num = np.sum((u_common - mean_u) * (v_common - mean_v))
            den = np.sqrt(np.sum((u_common - mean_u)**2)) * np.sqrt(np.sum((v_common - mean_v)**2))

            sim = (num / den) if den != 0 else 0

        elif sim_measure == "cos":
            num = np.sum(u_common * v_common)
            den = np.sqrt(np.sum(u_common**2)) * np.sqrt(np.sum(v_common**2))
            sim = (num / den) if den != 0 else 0
        else:
            sim = 0

        similarities[other_id] = sim

    sorted_neighbors = sorted(similarities.items(), key=lambda x: x[1], reverse=True)
    top_k_neighbors = [(v_id, sim) for v_id, sim in sorted_neighbors if sim > 0][:K]

    predictions = {}
    unrated_items = target_ratings[target_ratings == 0].index

    for item in unrated_items:
        valid_neighbors = [(v_id, sim) for v_id, sim in top_k_neighbors if data.loc[v_id, item] > 0]

        if not valid_neighbors:
            pred = mean_target
        else:
            numerator = 0
            denominator = 0

            for v_id, sim in valid_neighbors:
                v_ratings = data.loc[v_id]
                v_non_zero = v_ratings[v_ratings > 0]
                mean_v = v_non_zero.mean() if len(v_non_zero) > 0 else 0

                numerator += sim * (v_ratings[item] - mean_v)
                denominator += abs(sim)

            if denominator == 0:
                pred = mean_target
            else:
                pred = mean_target + (numerator / denominator)

        pred = max(1.0, min(5.0, pred))
        if is_rounded:
            pred = round(pred)

        predictions[item] = pred

    result_vector = pd.Series(0.0, index=target_ratings.index)

    for item, pred in predictions.items():
        result_vector[item] = pred

    return result_vector
```

```python
def CF_userbased(u_id, K, data, is_rounded=False, sim_measure="pearson"):
    if u_id not in data.index:
        raise ValueError(f"L'utilisateur {u_id} n'existe pas dans la matrice d'utilité.")

    target_ratings = data.loc[u_id]

    # Ignore les 0 et les -1 pour calculer la moyenne de l'utilisateur cible
    non_zero_target = target_ratings[target_ratings > 0]
    mean_target = non_zero_target.mean() if len(non_zero_target) > 0 else 0

    similarities = {}

    for other_id in data.index:
        if other_id == u_id:
            continue
        other_ratings = data.loc[other_id]

        # common_mask ne prend que les items notés par les deux (> 0)
        common_mask = (target_ratings > 0) & (other_ratings > 0)
        if not common_mask.any():
            continue

        u_common = target_ratings[common_mask]
        v_common = other_ratings[common_mask]

        if sim_measure == "pearson":
            mean_u = u_common.mean()
            mean_v = v_common.mean()

            num = np.sum((u_common - mean_u) * (v_common - mean_v))
            den = np.sqrt(np.sum((u_common - mean_u)**2)) * np.sqrt(np.sum((v_common - mean_v)**2))

            sim = (num / den) if den != 0 else 0

        elif sim_measure == "cos":
            num = np.sum(u_common * v_common)
            den = np.sqrt(np.sum(u_common**2)) * np.sqrt(np.sum(v_common**2))
            sim = (num / den) if den != 0 else 0
        else:
            sim = 0

        similarities[other_id] = sim

    sorted_neighbors = sorted(similarities.items(), key=lambda x: x[1], reverse=True)
    top_k_neighbors = [(v_id, sim) for v_id, sim in sorted_neighbors if sim > 0][:K]

    predictions = {}
    # Prend en compte à la fois 0 et -1 comme des items à prédire
    unrated_items = target_ratings[target_ratings <= 0].index

    for item in unrated_items:
        valid_neighbors = [(v_id, sim) for v_id, sim in top_k_neighbors if data.loc[v_id, item] > 0]

        if not valid_neighbors:
            pred = mean_target
        else:
            numerator = 0
            denominator = 0

            for v_id, sim in valid_neighbors:
                v_ratings = data.loc[v_id]
                v_non_zero = v_ratings[v_ratings > 0]
                mean_v = v_non_zero.mean() if len(v_non_zero) > 0 else 0

                numerator += sim * (v_ratings[item] - mean_v)
                denominator += abs(sim)

            if denominator == 0:
                pred = mean_target
            else:
                pred = mean_target + (numerator / denominator)

        pred = max(1.0, min(5.0, pred))
        if is_rounded:
            pred = round(pred)

        predictions[item] = pred

    result_vector = pd.Series(0.0, index=target_ratings.index)

    for item, pred in predictions.items():
        result_vector[item] = pred

    return result_vector
```

```python
import numpy as np
import pandas as pd

def create_u_test(u_vec):

    u_test = u_vec.copy()
    known_indices = u_test[u_test > 0].index

    if len(known_indices) == 0:
        return u_test

    # Nombre de valeurs à cacher (50% des notes connues)
    n_hide = int(np.ceil(len(known_indices) * 0.5))

    # Sélection aléatoire des indices à cacher
    hidden_indices = np.random.choice(known_indices, size=n_hide, replace=False)

    # Remplacement par -1
    u_test[hidden_indices] = -1
    return u_test
```

```python
import time
import numpy as np
import pandas as pd

def evaluate_cf_cross_validation(utility_matrix, K_neighbors=2, n_folds=5, sim_measure="pearson"):
    users = utility_matrix.index
    shuffled_users = np.random.permutation(users)
    folds = np.array_split(shuffled_users, n_folds)

    fold_rmses = []
    fold_maes = []
    fit_times = []
    test_times = []

    for fold_idx, test_users in enumerate(folds):
        squared_errors = []
        absolute_errors = []
        eval_matrix = utility_matrix.copy()

        # 1. TEMPS DE FIT (Préparation / Masquage)
        start_fit = time.perf_counter()
        for u_id in test_users:
            original_vec = utility_matrix.loc[u_id]
            u_test = create_u_test(original_vec)
            eval_matrix.loc[u_id] = u_test
        end_fit = time.perf_counter()
        fit_times.append(end_fit - start_fit)

        # 2. TEMPS DE TEST (Prédiction et Erreur)
        start_test = time.perf_counter()
        for u_id in test_users:
            original_vec = utility_matrix.loc[u_id]

            pred_vector = CF_userbased(
                u_id=u_id,
                K=K_neighbors,
                data=eval_matrix,
                is_rounded=False,
                sim_measure=sim_measure
            )

            hidden_items = original_vec[eval_matrix.loc[u_id] == -1].index

            for item in hidden_items:
                true_val = original_vec[item]
                pred_val = pred_vector[item]
                squared_errors.append((true_val - pred_val) ** 2)
                absolute_errors.append(abs(true_val - pred_val)) # Calcul de l'erreur absolue

        # Calcul des métriques pour ce fold
        if len(squared_errors) > 0:
            fold_rmses.append(np.sqrt(np.mean(squared_errors)))
            fold_maes.append(np.mean(absolute_errors))
        else:
            fold_rmses.append(0.0)
            fold_maes.append(0.0)

        end_test = time.perf_counter()
        test_times.append(end_test - start_test)

    # Création du DataFrame récapitulatif (exactement comme sur votre capture d'écran)
    fold_columns = [f"Fold {i+1}" for i in range(n_folds)]

    results_df = pd.DataFrame(
        {
            **{fold_columns[i]: [fold_rmses[i], fold_maes[i], fit_times[i], test_times[i]] for i in range(n_folds)},
            "Mean": [np.mean(fold_rmses), np.mean(fold_maes), np.mean(fit_times), np.mean(test_times)]
        },
        index=["RMSE (testset)", "MAE (testset)", "Fit time", "Test time"]
    )


    display(results_df.round(4))

    return fold_rmses, fold_maes, np.mean(fold_rmses), fit_times, test_times
```

```python
import pandas as pd
from surprise import Dataset, Reader

# 1. Transformer la matrice large en format long (stack)
stacked_df = utility_example.stack().reset_index()
stacked_df.columns = ['user', 'item', 'rating']

# 2. Filtrer les notes à 0 (éléments non notés)
stacked_df = stacked_df[stacked_df['rating'] > 0]

# 3. Charger le DataFrame dans le format attendu par Surprise
reader = Reader(rating_scale=(1, 5))
data = Dataset.load_from_df(stacked_df[['user', 'item', 'rating']], reader)
```

```python
from surprise.model_selection import KFold
from surprise import KNNWithMeans
from surprise import accuracy
import numpy as np

# Configuration des options de similarité
sim_options = {
    "name": "pearson",
    "user_based": True
}

# Initialisation de l'algorithme KNNWithMeans
algo = KNNWithMeans(sim_options=sim_options)

# Définition de l'itérateur de validation croisée (ici k = 5 folds)
kf = KFold(n_splits=5)

rmse_folds = []

for fold_idx, (trainset, testset) in enumerate(kf.split(data)):

    algo.fit(trainset)
    predictions = algo.test(testset)

    fold_rmse = accuracy.rmse(predictions, verbose=True)
    rmse_folds.append(fold_rmse)

mean_rmse = np.mean(rmse_folds)
print(f"\n--- RMSE Moyen sur les 5 folds : {mean_rmse:.4f} ---")
```
